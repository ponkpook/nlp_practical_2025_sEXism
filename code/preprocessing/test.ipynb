```python
import pandas as pd
from preprocessing import DataFramePreprocessor
from steps import strip_urls, lowercase, word_tokenizer

# 1. Define your pipeline once:
pipeline = DataFramePreprocessor([
    strip_urls,
    lowercase,
    word_tokenizer,
])

# 2. Simply pass your DataFrame + text column:
raw_df = pd.DataFrame({
    'tweet_text': [
        "Check this out! https://t.co/xyz",
        "I LOVE AI!!! #awesome"
    ]
})

processed_df = pipeline(raw_df, column='tweet_text')

print(processed_df)
```
